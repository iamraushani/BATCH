```python
import random

num=random.randint(1,100)
tries=0

while True:
    guessed=int(input("guess the number between 1 - 100"))
    
    tries+=1

    if guessed==num:
        print(f"congratulations you found your number in {tries} tries")
        break
    elif guessed>num:
        print("sorry you need to go lower")
    elif guessed<num:
        print("sorry you need to go a little upper")
```

```python
import random
import tkinter as tk
from tkinter import messagebox

cscore = 0
hscore = 0

# Create window
root = tk.Tk()
root.title("Stone Paper Scissors")
root.geometry("500x550")
root.config(bg="#1e1e2f")
root.resizable(False, False)

# Title
title = tk.Label(
    root,
    text="STONE PAPER SCISSORS",
    font=("Arial", 20, "bold"),
    bg="#1e1e2f",
    fg="white"
)
title.pack(pady=20)

# Instructions
instructions = tk.Label(
    root,
    text="First to 5 points wins the game!",
    font=("Arial", 12),
    bg="#1e1e2f",
    fg="lightgray"
)
instructions.pack()

# Score display
score_label = tk.Label(
    root,
    text="You: 0    Computer: 0",
    font=("Arial", 16, "bold"),
    bg="#1e1e2f",
    fg="#00ff99"
)
score_label.pack(pady=20)

# Result display
result_label = tk.Label(
    root,
    text="Choose your move!",
    font=("Arial", 14, "bold"),
    bg="#1e1e2f",
    fg="white"
)
result_label.pack(pady=10)

# Choices display
choice_label = tk.Label(
    root,
    text="You: ---   |   Computer: ---",
    font=("Arial", 12),
    bg="#1e1e2f",
    fg="orange"
)
choice_label.pack(pady=10)


def play(user):
    global cscore, hscore

    com = random.randint(1, 3)

    choices = {
        1: "Stone",
        2: "Paper",
        3: "Scissors"
    }

    choice_label.config(
        text=f"You: {choices[user]}   |   Computer: {choices[com]}"
    )

    if user == com:
        result_label.config(
            text="It's a Draw!",
            fg="yellow"
        )

    elif (
        (user == 1 and com == 3)
        or (user == 2 and com == 1)
        or (user == 3 and com == 2)
    ):
        hscore += 1

        result_label.config(
            text="You Won This Round! 🎉",
            fg="#00ff99"
        )

    else:
        cscore += 1

        result_label.config(
            text="Computer Won This Round!",
            fg="#ff5555"
        )

    score_label.config(
        text=f"You: {hscore}    Computer: {cscore}"
    )

    if hscore == 5:
        messagebox.showinfo(
            "Game Over",
            "Congratulations! You Won the Game! 🎉"
        )
        disable_buttons()

    elif cscore == 5:
        messagebox.showinfo(
            "Game Over",
            "Computer Won the Game! Better Luck Next Time!"
        )
        disable_buttons()


def disable_buttons():
    stone_button.config(state="disabled")
    paper_button.config(state="disabled")
    scissors_button.config(state="disabled")


def restart():
    global cscore, hscore

    cscore = 0
    hscore = 0

    score_label.config(text="You: 0    Computer: 0")
    result_label.config(
        text="Choose your move!",
        fg="white"
    )
    choice_label.config(
        text="You: ---   |   Computer: ---"
    )

    stone_button.config(state="normal")
    paper_button.config(state="normal")
    scissors_button.config(state="normal")


# Buttons
button_frame = tk.Frame(root, bg="#1e1e2f")
button_frame.pack(pady=25)

stone_button = tk.Button(
    button_frame,
    text="🪨 Stone",
    font=("Arial", 12, "bold"),
    width=10,
    bg="#444466",
    fg="white",
    command=lambda: play(1)
)
stone_button.grid(row=0, column=0, padx=5)

paper_button = tk.Button(
    button_frame,
    text="📄 Paper",
    font=("Arial", 12, "bold"),
    width=10,
    bg="#444466",
    fg="white",
    command=lambda: play(2)
)
paper_button.grid(row=0, column=1, padx=5)

scissors_button = tk.Button(
    button_frame,
    text="✂️ Scissors",
    font=("Arial", 12, "bold"),
    width=10,
    bg="#444466",
    fg="white",
    command=lambda: play(3)
)
scissors_button.grid(row=0, column=2, padx=5)

# Restart button
restart_button = tk.Button(
    root,
    text="🔄 Play Again",
    font=("Arial", 12, "bold"),
    width=15,
    bg="#00aa88",
    fg="white",
    command=restart
)
restart_button.pack(pady=15)

# Start the game
root.mainloop()
```
