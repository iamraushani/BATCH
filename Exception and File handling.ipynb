# **Class 16:Exception Handling**

Exception Handling
 When we run a program in Python, there are various exceptions that can be raised.
    Like - Syntax Error, Name Error, Zero Division Error etc.

 Exceptions are raised when the program is syntactically correct, but the code results in an error. This error does not stop the execution of the program; however, it changes the normal flow of the program.

 So you cannot handle the Syntax Error, but if there is no Syntax Error, you can handle them.


Exception Handling Functionalities
 Try - This will catch the exception if any and pass to -
  
 Except - Except will deal with the Exception you can have Custom Exceptions or Universal Catcher.

 Else - Else will be executed if no exception occurs or it won't be executed.

 Finally - This will definitely run no matter what happens.

 Raise - Raises a custom error as you need.

```python
a=int(input())
b=int(input())

try:
    print(a/b)  #b=0 ZeroDivisionError 

except Exception as err:
    print(f"sorry an error occured as {err}")

else:
    print("there is no errors")

finally:
    print("i will execute no matter what")
    
print(a+b)
```

```python
try:
    age=int(input())
    if age<18:
        raise Exception("you must be 18+")
    print("Access granted")
except Exception as e:
    print("Error:",e)
```

File Handling
