# IF-ELSE Logic Building

Python Logic Building
IF-ELSE Practice Set (1–10) – Descriptions

1. Compare Two Numbers
Take two user inputs and determine which number is greater — or if they're equal.

2. Greet by Gender (m/f)
Accept a gender input ('m' or 'f') and print a greeting like "Hello Sir" or "Hello Ma'am".
3. Gender with Case Handling
Make the gender check case-insensitive ('M', 'F', 'm', 'f' all valid). If input is invalid, print "Wrong Input".

4. Even or Odd Checker
Accept a number from the user and check whether it's even or odd using modulo (%).

5. Voting Eligibility
Input name and age. If age ≥ 18, print "Eligible to vote". If not, print how many years are left to become eligible.

6. Day Number to Day Name
Take an integer (1–7) and print the corresponding weekday (1 = Monday, 2 = Tuesday, ... 7 = Sunday). Handle invalid input too.

7. Greatest of Three Numbers
Accept three numbers and find the greatest one among them using nested if-else.
Also find if 2 are equal; also find if all are equal.

8. Leap Year Checker
Input a year and check if it's a leap year using proper rules (divisible by 4, not by 100 unless divisible by 400).

9. Shop Discount Calculator
Ask for purchase amount. Apply discounts based on thresholds: e.g., above ₹1000 → 10% off, above ₹5000 → 20% off. Print final bill. (You can also design a shop-like interface later.)

10. Vowel or Consonant
Accept a single alphabet character and check if it's a vowel (a, e, i, o, u).
Also, handle invalid characters.

```python
n1=float(input())
n2=float(input())

if n1>n2:
    print(f"{n1} is greater than {n2}")
elif n2>n1:
    print(f"{n2} is greater than {n1}")
else:
    print(f"{n1} is equal to {n2}")
```
